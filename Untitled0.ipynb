## Document Processing and Vector Store
We will now define the logic to load a PDF and store its embeddings in ChromaDB.

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from langchain.text_splitter import RecursiveCharacterTextSplitter

# Global variable for vector store
vector_store = None

def setup_knowledge_base(pdf_path):
    global vector_store
    loader = PyPDFLoader(pdf_path)
    documents = loader.load()

    text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=100)
    splits = text_splitter.split_documents(documents)

    vector_store = Chroma.from_documents(documents=splits, embedding=OpenAIEmbeddings())
    print("Knowledge base ready.")
```

## Core Functions
Implementing `retrieve_information` and `ask_ai` with Structured Outputs.

```python
from typing import Literal
from pydantic import BaseModel
from langchain_openai import ChatOpenAI
from gtts import gTTS
from IPython.display import Audio, display, Image

class QueryAnalysis(BaseModel):
    prompt: str
    format: Literal["text", "image", "audio"]

def retrieve_information(prompt: str) -> str:
    if vector_store is None:
        return "Error: Please load a PDF first."
    docs = vector_store.similarity_search(prompt, k=3)
    context = "\n".join([d.page_content for d in docs])

    llm = ChatOpenAI(model="gpt-4o")
    response = llm.invoke(f"Context: {context}\n\nQuestion: {prompt}\nAnswer:")
    return response.content

def generate_image(text_info: str):
    from openai import OpenAI
    client = OpenAI()
    response = client.images.generate(
        model="dall-e-3",
        prompt=f"A descriptive visualization based on this info: {text_info}",
        n=1,
        size="1024x1024"
    )
    return response.data[0].url

def ask_ai(question: str):
    llm = ChatOpenAI(model="gpt-4o")
    structured_llm = llm.with_structured_output(QueryAnalysis)

    # 1. Extract intent
    analysis = structured_llm.invoke(question)

    # 2. Retrieve info
    info = retrieve_information(analysis.prompt)

    # 3. Handle output format
    if analysis.format == "text":
        print(f"[TEXT RESPONSE]: {info}")
    elif analysis.format == "audio":
        tts = gTTS(info)
        tts.save("response.mp3")
        display(Audio("response.mp3", autoplay=True))
    elif analysis.format == "image":
        img_url = generate_image(info)
        display(Image(url=img_url))

    return info
```
